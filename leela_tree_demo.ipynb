# Leela Chess Zero Tree Extraction Demo

This notebook demonstrates how to initialize the Lc0 engine, run a search on a specific chess position, and extract/visualize the search tree.

**Note**: If the weights file is missing, the engine will use a `random` backend for demonstration purposes.

```python
import sys
import os
import chess
from IPython.display import display

# Ensure src is in path
sys.path.append(os.path.join(os.getcwd(), 'src'))

from leela_utils import get_leela_engine, run_search, extract_tree, visualize_tree
```

## 1. Initialize Engine

We point to the Lc0 binary and weights in the scratch directory.

```python
try:
    engine = get_leela_engine()
    print("Lc0 engine initialized successfully.")
except Exception as e:
    print(f"Error initializing engine: {e}")
```

## 2. Run Search and Extract Tree

We use a standard starting position and run a search with a limited node count.

```python
board = chess.Board()
nodes_to_search = 1000

print(f"Searching {nodes_to_search} nodes...")
search_info = run_search(board, engine, nodes=nodes_to_search)

print("Extracting tree...")
tree_root = extract_tree(board, search_info)
print(f"Root visits: {tree_root.visits}")
print(f"Top-level moves found: {[child.move for child in tree_root.children]}")
```

## 3. Visualize Tree

We use Graphviz to visualize the search tree.

```python
dot = visualize_tree(tree_root, max_depth=2)
display(dot)
```

```python
engine.quit()
```
